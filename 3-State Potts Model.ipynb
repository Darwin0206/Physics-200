#### Wolff Algorithm

```python
def new_state(L, q=3):
    return np.random.randint(0, q, size=(L, L))
def energy(spins, J=1.0):
    L = spins.shape[0]
    E = 0
    for x in range(L):
        for y in range(L):
            s = spins[x, y]
            # Nearest neighbors with periodic BC
            for dx, dy in [(1,0), (0,1)]:
                nx, ny = (x+dx) % L, (y+dy) % L
                if spins[nx, ny] == s:
                    E -= J
    return E
def magnetization(spins, q=3):
    L = spins.shape[0]
    #N = L*L
    # Represent spin states as unit vectors in complex plane
    angles = 2*np.pi*np.arange(q)/q
    vecs = np.exp(1j * angles)  # q complex unit vectors
    M = np.sum(vecs[spins])     # map spins to complex roots of unity
    return np.abs(M)
def wolff_cluster(spins, T, J=1.0, q=3):
    L = spins.shape[0]
    beta = 1.0 / T

    # Pick random seed
    i, j = randint(0, L, size=2)
    seed_state = spins[i, j]

    # Pick a random new spin state (different from seed_state)
    new_state_val = randint(0, q-1)
    if new_state_val >= seed_state:
        new_state_val += 1

    # Bond probability: p = 1 - exp(-beta*J)
    p = -np.expm1(-beta * J)

    cluster = {(i,j)}
    pocket = {(i,j)}

    while pocket:
        x, y = pocket.pop()
        for dx, dy in [(1,0), (-1,0), (0,1), (0,-1)]:
            nx, ny = (x+dx) % L, (y+dy) % L
            if (nx, ny) not in cluster and spins[nx,ny] == seed_state:
                if random() < p:
                    cluster.add((nx,ny))
                    pocket.add((nx,ny))

    # Flip cluster spins to new state
    for x, y in cluster:
        spins[x, y] = new_state_val
    return spins
```

```python
def run_simulation(L, T, sweeps=1000000, J=1.0, q=3, burn_frac=0.3):
    spins = new_state(L, q)
    E_arr, M_arr = [], []
    for _ in range(sweeps):
        spins = wolff_cluster(spins, T, J, q)
        E_arr.append(energy(spins, J))
        M_arr.append(magnetization(spins, q))
    # Burn-in removal
    burn_idx = int(burn_frac * sweeps)
    return np.array(M_arr[burn_idx:]), np.array(E_arr[burn_idx:])
```

```python
L = 16
T = 1

M_Wolff, E_Wolff = run_simulation(L, T)

# Magnetization plot
plt.figure(figsize=(12, 5))
plt.plot(M_Wolff, marker=',', linestyle='None', label='|Magnetization| (Metropolis)')
plt.title("Magnetization over Time")
plt.xlabel("Sweep")
plt.ylabel("Total Magnetization")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Energy plot
plt.figure(figsize=(12, 5))
plt.plot(E_Wolff, marker=',', linestyle='None', color='orange', label='Energy (Metropolis)')
plt.title("Energy over Time")
plt.xlabel("Sweep")
plt.ylabel("Total Energy")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Potts (q=3) Wolff implementation ---
def wolff_potts_step(lattice, beta, q=3):
    L = lattice.shape[0]
    # pick random seed site
    x, y = np.random.randint(0, L), np.random.randint(0, L) #returns an integer in [0, q-2]
    state = lattice[x, y]
    # choose a new state different from the seed state
    new_state = np.random.randint(0, q-1)
    if new_state >= state: # If that value is > or equal to seed state, adding 1 shifts it up by one skipping the seed value 
        new_state += 1
    
    # FK bond probability
    p_add = 1.0 - np.exp(-beta) #This is the Fortuin–Kasteleyn bond addition probability for the Potts model with coupling J=1. If the Hamiltonian uses -J δ(s_i,s_j), and J=1, then K = beta*J = beta, and p = 1 - e^{-K}.
    
    # cluster growth via stack (periodic boundary conditions)
    stack = [(x, y)]
    in_cluster = np.zeros((L, L), dtype=bool) # marking which sites are already in the cluster
    in_cluster[x, y] = True
    
    while stack: # Depth-first expansion: pop a site, check its neighbours 
        cx, cy = stack.pop()
        for dx, dy in ((1,0),(-1,0),(0,1),(0,-1)):
            nx, ny = (cx + dx) % L, (cy + dy) % L
            if (not in_cluster[nx, ny]) and lattice[nx, ny] == state: # If a neighbor is not in the cluster yet and it has the same state as the seed then add it with probability p_add
                if np.random.random() < p_add:
                    in_cluster[nx, ny] = True
                    stack.append((nx, ny)) # Push onto the stack to continue growth
    # flip the cluster
    lattice[in_cluster] = new_state # All sites that are in the cluster are set to cluster flip
    return lattice

def magnetization(lattice, q=3):
    # Represent each state as a root of unity on the complex plane and take |mean|
    angles = 2 * np.pi * lattice / q
    mx = np.cos(angles).sum() # x and y components of the vector sum
    my = np.sin(angles).sum()
    return np.sqrt(mx*mx + my*my) / lattice.size # magnitude of the total vector sum

def run_at_temperature(L, T, q=3, n_meas=10000, equil_frac=0.3, seed=None):
    if seed is not None:
        np.random.seed(seed) # Optionally seed Numpy RNG for reproducibility
    beta = 1.0 / T
    # hot start
    lattice = np.random.randint(0, q, size=(L, L)) # random initialization of the lattice with uniform states 
    n_equil = int(equil_frac * n_meas)  # number of equilibration steps
    
    # Equilibration (no measurements)
    for _ in range(n_equil): # Each equilibration step is one Wolff update
        lattice = wolff_potts_step(lattice, beta, q)
    
    # Measurements
    ms = np.empty(n_meas - n_equil, dtype=float) # after equilibration, this performs Wolff steps to measure only the magnetization after equilibration=
    for t in range(n_meas - n_equil):
        lattice = wolff_potts_step(lattice, beta, q)
        ms[t] = magnetization(lattice, q)
    return ms

# --- Run a small demo (fast) ---
L = 12
q = 3
# Critical temperature for 2D ferromagnetic q=3 Potts on the square lattice: Tc = 1 / ln(1 + sqrt(3))
Tc = 1.0 / np.log(1.0 + np.sqrt(3.0))
temps = np.linspace(0.6*Tc, 1.6*Tc, 10)

avg_m = []
for T in temps:
    ms = run_at_temperature(L, T, q=q, n_meas=12000, equil_frac=0.30, seed=123)
    avg_m.append(ms.mean())

# Plot |M| vs T
plt.figure(figsize=(6,4))
plt.plot(temps, avg_m, marker='o')
plt.axvline(Tc, linestyle='--')
plt.xlabel("Temperature T")
plt.ylabel("Magnetization |M|")
plt.title(f"3-state Potts (L={L}) — Wolff cluster; |M| vs T")
plt.grid(True)
plt.show()

# Also print the temperatures and their average magnetizations for reference
for T, m in zip(temps, avg_m):
    print(f"T={T:.4f},  <|M|>={m:.4f}")
```

```python
import numpy as np

def new_state(L, q=3):
    return np.random.randint(0, q, size=(L, L))
def total_energy(spins, J=1.0):
    L = spins.shape[0]
    E = 0.0

    for x in range(L):
        for y in range(L):
            s = spins[x, y]
            if spins[(x+1)%L, y] == s:
                E -= J
            if spins[x, (y+1)%L] == s:
                E -= J

    return E
def magnetization_counts(spins, q=3):
    counts = np.bincount(spins.ravel(), minlength=q)
    return counts
def wolff_cluster(spins, beta, counts, J=1.0, q=3):

    L = spins.shape[0]

    visited = np.zeros((L,L), dtype=bool)

    # random seed
    x0,y0 = np.random.randint(0,L,size=2)

    old_state = spins[x0,y0]

    # choose new state
    new_state_val = np.random.randint(0,q-1)
    if new_state_val >= old_state:
        new_state_val += 1

    p = 1 - np.exp(-beta*J)

    stack=[(x0,y0)]
    visited[x0,y0]=True

    cluster=[]

    while stack:

        x,y = stack.pop()
        cluster.append((x,y))

        for dx,dy in [(1,0),(-1,0),(0,1),(0,-1)]:

            nx=(x+dx)%L
            ny=(y+dy)%L

            if spins[nx,ny]==old_state and not visited[nx,ny]:

                if np.random.rand()<p:
                    visited[nx,ny]=True
                    stack.append((nx,ny))

    cluster_size=len(cluster)

    for x,y in cluster:
        spins[x,y]=new_state_val

    counts[old_state]-=cluster_size
    counts[new_state_val]+=cluster_size

    return spins, counts
def run_simulation(L, T, sweeps=10**5, J=1.0, q=3,
                   burn_frac=0.3, sample_interval=1):

    beta = 1.0/T

    spins = new_state(L,q)

    counts = magnetization_counts(spins,q)

    burn_steps=int(burn_frac*sweeps)

    E_sum=E2_sum=0.0
    M_sum=M2_sum=M4_sum=0.0

    n_samples=0

    N=L*L

    for step in range(sweeps):

        spins,counts = wolff_cluster(spins,beta,counts,J,q)

        if step>=burn_steps and (step-burn_steps)%sample_interval==0:

            E = total_energy(spins,J)

            m_max = np.max(counts)

            M = (q*m_max - N)/(q-1)

            E_sum += E
            E2_sum += E*E

            M_sum += M
            M2_sum += M*M
            M4_sum += M**4

            n_samples+=1

    E_avg=E_sum/n_samples
    E2_avg=E2_sum/n_samples

    M_avg=M_sum/n_samples
    M2_avg=M2_sum/n_samples
    M4_avg=M4_sum/n_samples

    M_per_site = M_avg/N

    chi_M = (M2_avg - M_avg**2)/(N*T)

    C_v = (E2_avg - E_avg**2)/(N*T**2)

    U = 1 - M4_avg/(3*M2_avg**2)

    return M_per_site, chi_M, C_v, E_avg/N, U
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
from scipy.interpolate import interp1d

# -----------------------
# Helper: build combined (x,y) for a choice of params
# -----------------------
def make_rescaled_data(L_values, T_values, results, Tc, nu, g_over_nu):
    all_x = []
    all_y = []
    labels = []
    for L in L_values:
        chi = np.array(results[L]["chi_M"], dtype=float)
        T = np.array(T_values, dtype=float)
        # remove NaNs / invalid
        mask = np.isfinite(chi) & np.isfinite(T)
        if np.sum(mask) < 3:
            continue
        chi = chi[mask]; T = T[mask]
        x = (T - Tc) * (L**(1.0/nu))               # scaling variable
        y = chi * (L**(-g_over_nu))               # scaled observable
        all_x.append(x)
        all_y.append(y)
        labels.append(np.full_like(x, L))
    if len(all_x) == 0:
        return None, None, None
    return all_x, all_y, labels

# -----------------------
# Collapse quality function
# Implementation: build a master curve by binning x from all datasets,
# compute mean y in each bin, compute weighted squared deviations.
# Lower score = better collapse.
# -----------------------
def collapse_score(params, L_values, T_values, results, nbins=60):
    """
    params = [Tc, nu, g_over_nu]
    """
    Tc, nu, g_over_nu = params
    # enforce sensible bounds implicitly by returning large cost
    if not (0.0 < nu < 5.0 and -5.0 < g_over_nu < 5.0 and np.isfinite(Tc)):
        return 1e50

    all_x, all_y, labels = make_rescaled_data(L_values, T_values, results, Tc, nu, g_over_nu)
    if all_x is None:
        return 1e50

    # concatenate
    xcat = np.concatenate(all_x)
    ycat = np.concatenate(all_y)

    # require enough points
    if xcat.size < 10:
        return 1e50

    # bin the x-range
    xmin, xmax = np.percentile(xcat, [1, 99])
    if xmax <= xmin:
        return 1e50
    bins = np.linspace(xmin, xmax, nbins+1)
    bin_centers = 0.5*(bins[:-1] + bins[1:])
    master_y = np.zeros_like(bin_centers)
    counts = np.zeros_like(bin_centers)

    # compute mean y in each bin (master curve)
    inds = np.digitize(xcat, bins) - 1
    for i in range(len(bin_centers)):
        mask = inds == i
        if np.sum(mask) > 0:
            master_y[i] = np.mean(ycat[mask])
            counts[i] = np.sum(mask)
        else:
            master_y[i] = np.nan

    # Interpolate master curve to x positions and compute residuals only where master defined
    valid = np.isfinite(master_y) & (counts >= 3)
    if np.sum(valid) < max(3, nbins//10):
        return 1e50
    interp = interp1d(bin_centers[valid], master_y[valid], bounds_error=False, fill_value=np.nan)

    y_master_at_x = interp(xcat)
    mask_ok = np.isfinite(y_master_at_x)
    if np.sum(mask_ok) < 10:
        return 1e50

    # normalized mean squared deviation: sum((y - y_master)^2) / sum(y_master^2)
    resid = ycat[mask_ok] - y_master_at_x[mask_ok]
    denom = np.sum(y_master_at_x[mask_ok]**2) + 1e-16
    score = np.sum(resid**2) / denom
    return score

# -----------------------
# Wrapper to run fit and plot
# -----------------------
def fit_and_plot_collapse(L_values, T_values, results,
                          Tc_guess=None, nu_guess=0.8, g_over_nu_guess=1.7,
                          bounds=None, nbins=60, method='Nelder-Mead', plot=True):
    L_values = np.array(L_values, dtype=float)
    # initial Tc guess: use largest-L peak if not provided
    if Tc_guess is None:
        # crude estimate: peak T for largest L
        Lmax = max(L_values)
        chi = np.array(results[Lmax]["chi_M"], dtype=float)
        Tc_guess = T_values[np.nanargmax(chi)]

    x0 = [Tc_guess, nu_guess, g_over_nu_guess]
    print("Initial guess:", x0)

    # minimize collapse score
    res = minimize(lambda p: collapse_score(p, L_values, T_values, results, nbins=nbins),
                   x0, method=method,
                   options={'maxiter': 2000, 'disp': True})
    Tc_fit, nu_fit, gnu_fit = res.x
    score = res.fun
    print("Fitted:", {"Tc":Tc_fit, "nu":nu_fit, "gamma_over_nu":gnu_fit, "score":score})

    # produce the collapsed plot
    if plot:
        all_x, all_y, labels = make_rescaled_data(L_values, T_values, results, Tc_fit, nu_fit, gnu_fit)
        plt.figure(figsize=(7,5))
        # plot each L in different color
        cmap = plt.get_cmap("tab10")
        for i, L in enumerate(L_values):
            if i >= len(all_x): break
            x = all_x[i]
            y = all_y[i]
            # sort for nicer lines
            idx = np.argsort(x)
            plt.plot(x[idx], y[idx], 'o-', ms=4, alpha=0.9, label=f"L={int(L)}", color=cmap(i % 10))

        # draw master curve (from bins)
        xcat = np.concatenate(all_x); ycat = np.concatenate(all_y)
        xmin, xmax = np.percentile(xcat, [1,99])
        bins = np.linspace(xmin, xmax, nbins+1)
        bin_centers = 0.5*(bins[:-1]+bins[1:])
        master_y = np.zeros_like(bin_centers)+np.nan
        inds = np.digitize(xcat, bins) - 1
        for i in range(len(bin_centers)):
            mask = inds == i
            if np.sum(mask) >= 3:
                master_y[i] = np.mean(ycat[mask])
        valid = np.isfinite(master_y)
        plt.plot(bin_centers[valid], master_y[valid], 'k--', lw=2, label='master curve')

        plt.xlabel(r'$(T - T_c)L^{1/\nu}$')
        plt.ylabel(r'$\chi L^{-\gamma/\nu}$')
        plt.title(f'Collapse: Tc={Tc_fit:.4f}, nu={nu_fit:.4f}, gamma/nu={gnu_fit:.4f}')
        plt.legend()
        plt.grid(True)
        plt.show()

    return {"Tc":Tc_fit, "nu":nu_fit, "gamma_over_nu":gnu_fit, "score":score, "optim_res":res}

# -----------------------
# Usage example:
# -----------------------
result = fit_and_plot_collapse(L_values, T_values, results,
                                Tc_guess=None, nu_guess=0.8, g_over_nu_guess=1.7)
print(result)

```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

def fss_collapse(results, L_values, T_values, Tc, nu, gamma_over_nu):
    collapsed_data = []
    for L in L_values:
        chi = np.array(results[L]["chi_M"])
        t_scaled = (T_values - Tc) * L**(1/nu)
        chi_scaled = chi * L**(-gamma_over_nu)
        collapsed_data.append((t_scaled, chi_scaled))
    return collapsed_data

def collapse_error(params, results, L_values, T_values):
    Tc, nu, gamma_over_nu = params
    data = fss_collapse(results, L_values, T_values, Tc, nu, gamma_over_nu)
    # Combine all datasets and measure vertical spread
    all_x = np.concatenate([x for x, _ in data])
    all_y = np.concatenate([y for _, y in data])
    bins = np.linspace(np.min(all_x), np.max(all_x), 50)
    bin_means = np.zeros_like(bins)
    for i in range(len(bins)-1):
        mask = (all_x >= bins[i]) & (all_x < bins[i+1])
        if np.any(mask):
            bin_means[i] = np.mean(all_y[mask])
    # Compute mean squared deviation from bin averages
    err = 0
    for x, y in data:
        for i in range(len(bins)-1):
            mask = (x >= bins[i]) & (x < bins[i+1])
            if np.any(mask):
                err += np.sum((y[mask] - bin_means[i])**2)
    return err / len(all_y)

def fit_and_plot_collapse(L_values, T_values, results, Tc_guess=1.0, nu_guess=1.0, gamma_over_nu_guess=1.75):
    p0 = [Tc_guess, nu_guess, gamma_over_nu_guess]
    bounds = [(0.9, 1.1), (0.1, 5.0), (0.1, 3.0)]
    res = minimize(collapse_error, p0, args=(results, L_values, T_values), bounds=bounds)
    Tc_opt, nu_opt, gamma_over_nu_opt = res.x

    print(f"Optimized parameters:")
    print(f"  Tc          = {Tc_opt:.5f}")
    print(f"  nu          = {nu_opt:.4f}")
    print(f"  gamma/nu    = {gamma_over_nu_opt:.4f}")
    print(f"  Collapse error = {res.fun:.5e}")

    data = fss_collapse(results, L_values, T_values, Tc_opt, nu_opt, gamma_over_nu_opt)
    plt.figure(figsize=(6,5))
    for L, (x, y) in zip(L_values, data):
        plt.plot(x, y, 'o-', label=f"L={L}")
    plt.xlabel(r"$(T - T_c)L^{1/\nu}$", fontsize=12)
    plt.ylabel(r"$\chi L^{-\gamma/\nu}$", fontsize=12)
    plt.title("Finite-Size Scaling Collapse of Susceptibility", fontsize=13)
    plt.legend()
    plt.grid(True)
    plt.show()
    return Tc_opt, nu_opt, gamma_over_nu_opt
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.random import randint, random

# ---------------------------
# Parameters (tweak as needed)
# ---------------------------
J = 1.0
q = 3

# start small for testing; increase for production
sweeps = 30000
burn_frac = 0.25

L_values = [6, 8, 10, 12]            # sizes (include pairs like L and 2L if possible)
T_values = np.linspace(0.7, 1.3, 25) # temperature grid around Tc ~ 0.995

# ---------------------------
# Model / Wolff functions
# ---------------------------
def new_state(L, q=3):
    return np.random.randint(0, q, size=(L, L))

def E(spins, J=1.0):
    L = spins.shape[0]
    energy_val = 0
    for x in range(L):
        for y in range(L):
            s = spins[x, y]
            # right and down neighbors to avoid double counting
            if spins[(x+1) % L, y] == s:
                energy_val -= J
            if spins[x, (y+1) % L] == s:
                energy_val -= J
    return energy_val

def magnetization(spins, q=3):
    # returns total |M| (not per-site); consistent with susceptibility formula used below
    angles = 2 * np.pi * np.arange(q) / q
    vecs = np.exp(1j * angles)
    M = np.sum(vecs[spins])
    return np.abs(M)  # total magnitude

def wolff_cluster(spins, T, J=1.0, q=3):
    L = spins.shape[0]
    beta = 1.0 / T
    i, j = randint(0, L, size=2)
    seed_state = spins[i, j]

    # choose new state != seed_state
    new_state_val = randint(0, q-1)
    if new_state_val >= seed_state:
        new_state_val += 1

    p = 1 - np.exp(-beta * J)

    cluster = {(i, j)}
    pocket = {(i, j)}

    while pocket:
        x, y = pocket.pop()
        for dx, dy in [(1,0), (-1,0), (0,1), (0,-1)]:
            nx, ny = (x + dx) % L, (y + dy) % L
            if (nx, ny) not in cluster and spins[nx, ny] == seed_state:
                if random() < p:
                    cluster.add((nx, ny))
                    pocket.add((nx, ny))

    for (x, y) in cluster:
        spins[x, y] = new_state_val
    return spins

# ---------------------------
# Observables & helpers
# ---------------------------
def binder_from_samples(M_arr):
    M2 = np.mean(M_arr**2)
    M4 = np.mean(M_arr**4)
    # guard against tiny rounding issues
    if M2 <= 0:
        return np.nan
    return 1.0 - M4 / (3.0 * (M2**2))

def run_simulation(L, T, sweeps=sweeps, J=1.0, q=3, burn_frac=burn_frac):
    """Return per-site mean magnetization, susceptibility per-site, U4 (Binder),
       and internal arrays M_arr, E_arr for potential further analysis."""
    N = L**2
    spins = new_state(L, q)
    burn_steps = int(burn_frac * sweeps)

    M_arr = []
    E_arr = []

    for step in range(sweeps):
        spins = wolff_cluster(spins, T, J, q)
        if step >= burn_steps:
            m = magnetization(spins, q)  # total |M|
            M_arr.append(m)
            E_arr.append(E(spins, J))

    M_arr = np.array(M_arr, dtype=float)
    E_arr = np.array(E_arr, dtype=float)

    if M_arr.size == 0:
        raise RuntimeError("No measurements collected; reduce burn_frac or increase sweeps")

    # Averages: note M_arr is total |M|
    M_avg_total = np.mean(M_arr)
    M_avg_per_site = M_avg_total / N

    M2 = np.mean(M_arr**2)
    # susceptibility per site: (⟨M^2⟩ - ⟨M⟩^2) / (N * T)
    chi_per_site = (M2 - M_avg_total**2) / (N * T)

    U4 = binder_from_samples(M_arr)

    return M_avg_per_site, chi_per_site, U4, M_arr, E_arr

# ---------------------------
# Run sims and store results
# ---------------------------
results = {L: {"M_x": [], "chi_M": [], "U4": [], "M_series": []} for L in L_values}

print("Starting simulations (this may take time)...")
for L in L_values:
    print(f" L={L}")
    for T in T_values:
        M_avg, chi_ps, U4_val, M_arr, E_arr = run_simulation(L, T, sweeps=sweeps, J=J, q=q)
        results[L]["M_x"].append(M_avg)
        results[L]["chi_M"].append(chi_ps)
        results[L]["U4"].append(U4_val)
        results[L]["M_series"].append(M_arr)  # optional: store full M series for each T

# Convert lists to arrays
for L in L_values:
    results[L]["M_x"] = np.array(results[L]["M_x"])
    results[L]["chi_M"] = np.array(results[L]["chi_M"])
    results[L]["U4"] = np.array(results[L]["U4"])

# ---------------------------
# Binder crossing & quotient method for 1/nu
# ---------------------------
def crossing_temperature(T, U_L, U_sL):
    D = U_L - U_sL
    # find indices where sign changes
    s = np.sign(D[:-1]) * np.sign(D[1:])
    idx = np.where(s < 0)[0]
    if len(idx) == 0:
        return None
    # pick the first crossing (or customize)
    i = idx[0]
    t0, t1 = T[i], T[i+1]
    d0, d1 = D[i], D[i+1]
    # linear interpolation of D(t)=0
    return t0 - d0 * (t1 - t0) / (d1 - d0)

def binder_slope_at(T, U, T0):
    i = np.searchsorted(T, T0)
    i = np.clip(i, 1, len(T)-2)
    dU = U[i+1] - U[i-1]
    dT = T[i+1] - T[i-1]
    return dU / dT

def inv_nu_from_quotient(SL, S_sL, s):
    return np.log(S_sL / SL) / np.log(s)

# compute crossings for pairs (L, 2L) when possible
pairs = [(L, 2*L) for L in L_values if 2*L in L_values]
inv_nu_estimates = []
crossings = []

plt.figure(figsize=(6,4))
for L in L_values:
    plt.plot(T_values, results[L]["U4"], marker='o', label=f"L={L}")
for (L1, L2) in pairs:
    T_cross = crossing_temperature(T_values, results[L1]["U4"], results[L2]["U4"])
    if T_cross is None:
        print(f"No crossing found for pair {L1},{L2} on current T grid.")
        continue
    crossings.append(T_cross)
    S1 = binder_slope_at(T_values, results[L1]["U4"], T_cross)
    S2 = binder_slope_at(T_values, results[L2]["U4"], T_cross)
    inv_nu = inv_nu_from_quotient(S1, S2, s=L2 / L1)
    inv_nu_estimates.append(inv_nu)
    plt.axvline(T_cross, color='gray', linestyle='--', alpha=0.6)
    print(f"Pair L={L1}, {L2}: crossing at T≈{T_cross:.5f}, 1/nu≈{inv_nu:.4f}")

plt.xlabel("Temperature T")
plt.ylabel("Binder U4")
plt.legend()
plt.title("Binder cumulant (U4) vs T and crossing estimates")
plt.grid(True)
plt.show()

if len(inv_nu_estimates) > 0:
    inv_nu_mean = np.mean(inv_nu_estimates)
    print(f"Binder-slope quotient: 1/nu (mean over pairs) ≈ {inv_nu_mean:.4f}, nu ≈ {1.0/inv_nu_mean:.4f}")
else:
    print("No inv_nu estimates from Binder slopes (no valid crossings).")

# ---------------------------
# Chi-peak method (shift of T*_chi) and gamma/nu from chi_max
# ---------------------------
def quadratic_peak_interpolation(x, y):
    """Given x and y (1D arrays) return refined peak location x_peak and y_peak.
       Uses quadratic fit around the maximum index (requires len>=3)."""
    imax = np.argmax(y)
    if imax == 0 or imax == len(y)-1 or len(y) < 3:
        # boundary or insufficient points: return coarse peak
        return x[imax], y[imax]
    # take three points
    x0, x1, x2 = x[imax-1], x[imax], x[imax+1]
    y0, y1, y2 = y[imax-1], y[imax], y[imax+1]
    # fit parabola y = a x^2 + b x + c -> vertex at -b/(2a)
    denom = (x0 - x1)*(x0 - x2)*(x1 - x2)
    if denom == 0:
        return x1, y1
    A = (x2*(y1 - y0) + x1*(y0 - y2) + x0*(y2 - y1)) / denom
    B = (x2**2*(y0 - y1) + x1**2*(y2 - y0) + x0**2*(y1 - y2)) / denom
    if A == 0:
        return x1, y1
    x_peak = -B / (2*A)
    y_peak = A*x_peak**2 + B*x_peak + (y0 - A*x0**2 - B*x0)  # compute using parabola
    return x_peak, y_peak

T_star_list = []
chi_max_list = []
Ls_for_peaks = []

for L in L_values:
    chi_arr = np.array(results[L]["chi_M"])
    # find refined peak
    T_peak, chi_peak = quadratic_peak_interpolation(T_values, chi_arr)
    T_star_list.append(T_peak)
    chi_max_list.append(chi_peak)
    Ls_for_peaks.append(L)

# Fit log(chi_max) vs log(L) -> slope = gamma/nu
logLs = np.log(np.array(Ls_for_peaks))
logChiMax = np.log(np.maximum(np.array(chi_max_list), 1e-16))
slope_gamma_over_nu, intercept = np.polyfit(logLs, logChiMax, 1)
gamma_over_nu = slope_gamma_over_nu
print(f"Fit chi_max ~ L^(gamma/nu): gamma/nu ≈ {gamma_over_nu:.4f}")

# Use Binder crossing mean as Tc estimate (if available)
if len(crossings) > 0:
    Tc_est = np.mean(crossings)  # simple average of pair crossings
    print(f"Using Tc_est from Binder crossings = {Tc_est:.6f}")
    # compute delta = T_star(L) - Tc_est
    deltas = np.array(T_star_list) - Tc_est
    # keep only positive deltas (above Tc)
    mask = deltas > 0
    if np.sum(mask) >= 2:
        slope_shift, intercept_shift = np.polyfit(np.log(np.array(Ls_for_peaks)[mask]), np.log(deltas[mask]), 1)
        # slope_shift = -1/nu
        inv_nu_shift = -slope_shift
        print(f"From chi-peak shift: 1/nu ≈ {inv_nu_shift:.4f},  nu ≈ {1.0/inv_nu_shift:.4f}")
    else:
        print("Not enough positive (T*-Tc) values to fit shift law; try different Tc estimate or larger L range.")
else:
    print("No Tc estimate from Binder crossings; cannot do chi-shift fit reliably.")

# ---------------------------
# Plot chi peaks vs L (log-log) and T_star vs L^-1/nu (if we have inv_nu)
# ---------------------------
plt.figure(figsize=(6,4))
plt.loglog(Ls_for_peaks, chi_max_list, 'o-')
plt.xlabel("L")
plt.ylabel("chi_max")
plt.title("chi peak vs L (log-log)")
plt.grid(True)
plt.show()

if len(crossings) > 0 and 'inv_nu_shift' in locals():
    # collapse plot of T_star vs L^{-1/nu}
    plt.figure(figsize=(6,4))
    plt.plot(1/np.array(Ls_for_peaks)**inv_nu_shift, T_star_list, 'o-')
    plt.xlabel(r"L$^{-1/\nu}$")
    plt.ylabel(r"T$^*_\chi$(L)")
    plt.title("Shift of chi peak vs L^{-1/nu}")
    plt.grid(True)
    plt.show()

print("Done.")
```

```python

def new_state(L, q=3):
    return np.random.randint(0, q, size=(L, L))
def E(spins, J=1.0):
    L = spins.shape[0]
    energy_val= 0
    for x in range(L):
        for y in range(L):
            s = spins[x, y]
            # Nearest neighbors with periodic BC
            for dx, dy in [(1,0), (0,1)]:
                nx, ny = (x+dx) % L, (y+dy) % L
                if spins[nx, ny] == s:
                    energy_val -= J
    return energy_val
def magnetization(spins, q=3):
    L = spins.shape[0]
    #N = L*L
    # Represent spin states as unit vectors in complex plane
    angles = 2*np.pi*np.arange(q)/q
    cos_vals = np.cos(angles)
    sin_vals = np.sin(angles)

    # Map spins directly
    Mx = np.sum(cos_vals[spins])
    My = np.sum(sin_vals[spins])
    M = np.sqrt(Mx**2 + My**2) 
    #M = Mx + 1j*My
    return M
def wolff_cluster(spins, T, J=1.0, q=3):
    L = spins.shape[0]
    beta = 1.0 / T

    # Pick random seed
    i, j = randint(0, L, size=2)
    seed_state = spins[i, j]

    # Pick a random new spin state (different from seed_state)
    new_state_val = randint(0, q-1)
    if new_state_val >= seed_state:
        new_state_val += 1

    # Bond probability: p = 1 - exp(-beta*J)
    p = 1 - np.exp(-beta * J)

    cluster = {(i,j)}
    pocket = {(i,j)}

    while pocket:
        x, y = pocket.pop()
        for dx, dy in [(1,0), (-1,0), (0,1), (0,-1)]:
            nx, ny = (x+dx) % L, (y+dy) % L
            if (nx, ny) not in cluster and spins[nx,ny] == seed_state:
                if random() < p:
                    cluster.add((nx,ny))
                    pocket.add((nx,ny))

    # Flip cluster spins to new state
    for x, y in cluster:
        spins[x, y] = new_state_val
    return spins
# --- Binder Cumulant Function ---
def binder_cumulant(M_arr, burn_frac=0.3):
    """
    Compute the Binder cumulant:
        U = 1 - <M^4> / (3 <M^2>^2)
    M_arr : array of magnetization values from the simulation
    burn_frac : fraction of data to discard for equilibration
    """
    burn_in_index = int(burn_frac * len(M_arr))
    M_trunc = np.array(M_arr[burn_in_index:])

    # Compute Binder cumulant
    M2 = np.mean(M_trunc**2)
    M4 = np.mean(M_trunc**4)
    U = 1 - (M4 / (3 * M2**2))
    return U

# --- Modified run_simulation to record M_arr ---
def run_simulation_binder(L, T, sweeps=10**6, J=1.0, q=3, burn_frac=0.3, sample_interval=1):
    spins = new_state(L, q)
    M_arr = []
    burn_steps = int(burn_frac * sweeps)

    for step in range(sweeps):
        spins = wolff_cluster(spins, T, J)
        if step >= burn_steps and (step - burn_steps) % sample_interval == 0:
            M_arr.append(magnetization(spins))
    
    return np.array(M_arr)

# --- Storage for Binder cumulant results ---
binder_results = {L: [] for L in L_values}

# --- Loop over system sizes and temperatures ---
for L in L_values:
    print(f"\nRunning Binder Cumulant simulations for L = {L}")
    for T in T_values:
        start_time = time.time()
        M_arr = run_simulation_binder(L, T, sweeps=sweeps, J=J, q=q, burn_frac=0.3)
        U = binder_cumulant(M_arr)
        binder_results[L].append(U)
        elapsed = time.time() - start_time
        print(f"  T = {T:.3f}, U = {U:.4f}, time = {elapsed/60:.2f} min")

# --- Plot Binder Cumulant vs Temperature ---
plt.figure(figsize=(7,5))
for L in L_values:
    plt.plot(T_values, binder_results[L], 'o-', label=f"L={L}")
plt.xlabel("Temperature $T$")
plt.ylabel("Binder Cumulant $U$")
plt.title("Binder Cumulant vs Temperature for 3-state Potts Model")
plt.legend()
plt.grid(True)
plt.show()
```
